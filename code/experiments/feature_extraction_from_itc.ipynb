```python
# Dimensions
channels = [f'Ch_{i}' for i in range(three.shape[0])]
frequencies = np.arange(1, 30)
time_points = np.linspace(-0.2, 9.5, three.shape[2])

# Create a MultiIndex
index = pd.MultiIndex.from_product([channels, frequencies, time_points],
                                   names=['Channel', 'Frequency', 'Time'])

# Flatten the data to fit into a DataFrame
flattened_data = three.flatten()

# Create the DataFrame
df = pd.DataFrame(flattened_data, index=index, columns=['ITC'])

# Reshape the DataFrame
three_df = df.reset_index().pivot_table(index=['Channel', 'Frequency'], columns='Time', values='ITC')
```
